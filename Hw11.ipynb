```python
from keras.datasets import imdb
from keras.utils import pad_sequences
from keras.models import Sequential
from keras.layers import Embedding, SimpleRNN, GRU, LSTM, Bidirectional, Dense, Dropout
from keras.callbacks import EarlyStopping
from keras.optimizers import Adam, RMSprop
```

```python
max_features = 10000
maxlen = 500
batch_size = 64
```

```python
(x_train, y_train), (x_test, y_test) = imdb.load_data(
    num_words=max_features
)

x_train = pad_sequences(x_train, maxlen=maxlen)
x_test = pad_sequences(x_test, maxlen=maxlen)
```

# RNN

```python
model = Sequential()
model.add(Embedding(max_features, 32, mask_zero=True))
model.add(SimpleRNN(64, dropout=0.2))
model.add(Dense(1, activation="sigmoid"))

optimizer = Adam(learning_rate=0.0005)

model.compile(
    optimizer=optimizer,
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

early_stopping = EarlyStopping(
    monitor="val_accuracy",
    patience=3,
    restore_best_weights=True
)

history = model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=batch_size,
    validation_split=0.2,
    callbacks=[early_stopping]
)
```

```python
test_loss, test_acc = model.evaluate(x_test, y_test)
print(f"Test Loss: {test_loss:.4f}")
print(f"Test Accuracy: {test_acc:.4f}")
```

# GRU

```python
(x_train, y_train), (x_test, y_test) = imdb.load_data(
    num_words=max_features
)

x_train = pad_sequences(
    x_train,
    maxlen=maxlen,
    padding="post"
)

x_test = pad_sequences(
    x_test,
    maxlen=maxlen,
    padding="post"
)
```

```python
model_gru = Sequential()
model_gru.add(Embedding(max_features, 32, mask_zero=True))
model_gru.add(GRU(64, dropout=0.2))
model_gru.add(Dense(1, activation="sigmoid"))

optimizer = RMSprop(learning_rate=0.0005)

model_gru.compile(
    optimizer=optimizer,
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

history_gru = model_gru.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=batch_size,
    validation_split=0.2,
    callbacks=[early_stopping]
)
```

```python
test_loss, test_acc = model_gru.evaluate(x_test, y_test)
print(f"Test Loss: {test_loss:.4f}")
print(f"Test Accuracy: {test_acc:.4f}")
```

# LSTM

```python
model_lstm = Sequential()
model_lstm.add(Embedding(max_features, 32, mask_zero=True))
model_lstm.add(LSTM(32))
model_lstm.add(Dense(1, activation="sigmoid"))


model_lstm.compile(
    optimizer="rmsprop",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

history_lstm = model_lstm.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=batch_size,
    validation_split=0.2,
    callbacks=[early_stopping]
)
```

```python
test_loss, test_acc = model_lstm.evaluate(x_test, y_test)
print(f"Test Loss: {test_loss:.4f}")
print(f"Test Accuracy: {test_acc:.4f}")
```

# BRNN

```python
model_brnn = Sequential()
model_brnn.add(Embedding(max_features, 32, mask_zero=True))
model_brnn.add(Bidirectional(GRU(64, dropout=0.2)))
model_brnn.add(Dense(1, activation="sigmoid"))

optimizer = RMSprop(learning_rate=0.0005)

model_brnn.compile(
    optimizer=optimizer,
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

history_brnn = model_brnn.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=batch_size,
    validation_split=0.2,
    callbacks=[early_stopping]
)
```

```python
test_loss, test_acc = model_brnn.evaluate(x_test, y_test)
print(f"Test Loss: {test_loss:.4f}")
print(f"Test Accuracy: {test_acc:.4f}")
```

# DRNN

```python
model_drnn = Sequential()
model_drnn.add(Embedding(max_features, 32, mask_zero=True))
model_drnn.add(GRU(64, return_sequences=True, dropout=0.2))
model_drnn.add(GRU(64))
model_drnn.add(Dense(1, activation="sigmoid"))

optimizer = RMSprop(learning_rate=0.0005)

model_drnn.compile(
    optimizer=optimizer,
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

history_drnn = model_drnn.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=batch_size,
    validation_split=0.2,
    callbacks=[early_stopping]
)
```

```python
test_loss, test_acc = model_drnn.evaluate(x_test, y_test)
print(f"Test Loss: {test_loss:.4f}")
print(f"Test Accuracy: {test_acc:.4f}")
```

# Висновок:

**SimpleRNN**


---


Найкраща Val accuracy: **82.28%**
Test accuracy: **86.48%**

SimpleRNN достатньо непогано розрізняє рецензії, але схильна до перенавчання: на початку експеременту точність на навчальній вибірці дорівнювала 99%, а на валідаційній близько 70%. Запровадження дропауту не покращило точність і вона упала до 64%, але зменшення кроку навчання до 0.0005 дало суттєві зміни і в рамках цього експеременту була досягнута найвища точність 82.28%.



**GRU**


---


Найкраща Val accuracy: **88.98%**
Test accuracy: **87.54%**

Ця архітектура показала добрі здібності класифікації даних.
Збільшення нейронів з 32 до 64 не вплинули на покращення валідаційної точності, хоча навчальна точність зросла. Оптимізатор RMSprop показав кращі результати точності ніж Adam, а зменшення кроку навчання допомогло знизити валідаційну помилку. Додавання дропауту допомогло частково контролювати актине перенавчання та наблизити значення навчальної та валідаційної точності один до одного.

**LSTM**


---

Найкраща Val accuracy: **88.66%**
Test accuracy: **87.54%**

Дана архітектура також непогано розрізняє рецензії, але в ході експерементів не перевищила GRU. Збільшення кількості нейронів з 32 до 64 не вплинуло на модель, так само як і зменшення навчального кроку, запровадження дропауту та вибір Adam замість RMSprop не дало корисних показників. Також був проведений експеремент із збільшенням нейронів та дропауту: 32 нейрони ⟶ 128, dropout 0.2 ⟶ 0.3, але покращень точності не відбулося. Ці дослідження дають зрозуміти, що  складна модель не є гарантовано найкращим варіантом, і було прийнято рішення залишити просту модель із 32 нейронами та дефолтним кроком навчання.

**BRNN**


---

Найкраща Val accuracy: **88.78%**
Test accuracy: **87.46%**

Спочатку було обрано просту рекурентну мережу Bidirectional(SimpleRN). Збільшенні нейронів з 32 до 64  підвищило валідаційну точність приблизно на 5%. Використання шаблону оптимізатора RMSprop + 0.0005 знову виявилось кращім варіантом. Останнім експерементом в рамках цієї архітектури було зміна звичайної рекурентної мережі на Bidirectional(GRU(64)) + RMSprop + 0.0005 + Dropout. Вибір зумовлений тим, що до цього моменту комбінація із GRU показувала найліпші результати, і серед усіх експеремнтів із Bidirectional це виявилась найвдаліша конфігурація, але тестова точність все ще не змогла перевищити точність звичайної GRU моделі.

**DRNN**


---

Найкраща Val accuracy: **88.80%**

Test accuracy: ***88.66%***≪≪≪≪

В цьому експеременті зберігається шаблон оптимізатора RMSprop + 0.0005 і більше приділяється уваги архітектурі.
Глибока рекурентна мережа показала найвищу в дослідженні тестову точність серед усіх розглянутих моделей. Два однакових шари GRU(64) сформували валідаційну точність 86% і додатковий дропаут збілишв її на 2% і зменшив помилку з 35% до 27%. Поєднання GRU(32) + LSTM(64) + Dropout зовсім несуттєво покращило точність, а валідаційна помилка навіть сбільшилась. Це знову доказ того, що складна мережа не гарантує високу точність.
